```python
# imports

from openai import OpenAI
from dotenv import load_dotenv
import os
from IPython.display import display, Markdown, update_display
```

```python
# constants

load_dotenv(override=True)
MODEL_GPT = 'gpt-4o-mini'
API_KEY = os.getenv('OPENAI_API_KEY')
MODEL_LLAMA = 'llama3.2'
BASE_API_URL = 'http://localhost:11434/v1'
```

```python
   
client = OpenAI(base_url=BASE_API_URL, api_key='ollama')

system_prompt = """
You are a technical coding assistant
You will receive a question
You should answer with a detailed explanation
Use markdown to format and emphasize important terms
"""

def get_messages_for(question):
    messages = [
        { 'role': 'system', 'content': system_prompt },
        { 'role': 'user', 'content': question }
    ]
    
    return messages

def get_answer_for(question, stream=False):
    messages = get_messages_for(question)
    response = client.chat.completions.create(
        model=MODEL_LLAMA,
        messages=messages,
        stream=stream
    )

    return response

def print_answer_for(question):
    response = get_answer_for(question)
    answer = response.choices[0].message.content

    display(Markdown(answer))

def print_answer_as_stream_for(question):
    stream = get_answer_for(question, stream=True)
    display_handle = display(Markdown(''), display_id=True)
    response =  ''

    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
print_answer_as_stream_for(question)
```

```python
# Get Llama 3.2 to answer
print_answer_for(question)
```
